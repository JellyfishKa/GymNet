Подготовка данных для классов:

Цель: извлечь landmarks, нормализовать, собрать скользящие окна и подготовить fold-сплиты.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold

CLASSES = ["ResistanceBand", "PushUps", "Squats", "RunInPlace"]
WINDOW_SIZE = 13
WINDOW_STEP = 3

DATA_ROOT = Path("../data/raw")
OUT_ROOT = Path("../data/processed")
OUT_ROOT.mkdir(parents=True, exist_ok=True)

print("Классы:", CLASSES)
print("Окно:", WINDOW_SIZE, "шаг:", WINDOW_STEP)
```

```python
def build_windows(sequence: np.ndarray, label: int, window_size: int, step: int):
    xs, ys = [], []
    if len(sequence) < window_size:
        return xs, ys
    for start in range(0, len(sequence) - window_size + 1, step):
        end = start + window_size
        xs.append(sequence[start:end])
        ys.append(label)
    return xs, ys

# Ожидаемая форма последовательности: [шаги_времени, число_признаков]
example = np.random.rand(40, 99).astype(np.float32)
x_win, y_win = build_windows(example, label=0, window_size=WINDOW_SIZE, step=WINDOW_STEP)
print("Окон:", len(x_win), "Пример меток:", y_win[:3])
```
